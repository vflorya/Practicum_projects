### Наличие пропусков в данных


```python
# Выводим количество пропущенных строк в датафрейме
new_games.isna().sum()
```

```python
# Подсчитываем процент строк с пропусками
new_games.isna().sum() / len(new_games) * 100
```

Изучили наличие пропущенных строк в датафрейме: в 'name'- 2 (0,01%) пустые строки, 'year_of_release' - 275 (1,6%) пустых строк, genre - 2 (0.01%) пустые строки, eu_sales - 6 (0.03%) пустых строк, jp_sales - 4 (0.02%) пусте строки, 'critic_score' - 8714 (51%) пустых строк, 'user_score' - 9268 (54,7%) пустые строки, 'rating' - 6871 (40,5%) пустых строк. Можно предположить, что критики и пользователи пропускали заполнение данных с оценкой, поэтому и много пустых значений в столбце с рейтингом.
- `name`: Игра без названия — это бесполезные данные, которые невозможно идентифицировать. Решение: игру без названия можно смело удалять. 
- `year_of_release`: Пропущено 275 значений,это критично для временного анализа. Заполняем медианным значением.
- `genre`: Количество пропусков незначительно (2 записи, меньше 1%). Анализ по жанрам (группировка, сравнение) невозможен без этой информации. Заполнение случайным жанром исказило бы статистику по жанрам.Решение: удаляем.
- `eu_sales`: Скорее всего, игры не продавались в этом регионе.Решение: заменить на ноль.
- `jp_sales`: Скорее всего, игры не продавались в этом регионе. Решение: заменить на ноль.
- `critic_score`: Удаление 51% данных неприемлемо — потеряется слишком много информации. Заполнение средним исказит реальное распределение оценок.Решение: оставить NaN, но создать отдельный столбец has_critic_score, указывающий, есть ли оценка.
- `user_score`: Заполнение средним создаст ложную корреляцию. Удаление 40% данных неприемлемо. Решение:оставить пропуски (NaN) + создать столбец-индикатор.
- `rating `: Для 40% игр не указан возрастной рейтинг. Решение: Категориальный признак, заполнить значением "Unknown".

```python
# Создаем копию для дальнейшей работы с данными 
games_clean = new_games.copy()
```

```python
# Удаляем игры без названия (их всего 2)
games_clean = games_clean.dropna(subset=['name'])
```

```python
# Удаляем жанр без названия (их всего 2)
games_clean = games_clean.dropna(subset=['genre'])
```

```python
# Заполним медианным годом (не хотим терять данные) пропуски в столце с годом
median_year = games_clean['year_of_release'].median()
games_clean['year_of_release'] = games_clean['year_of_release'].fillna(median_year).astype('Int64')
```

```python
# Оценку критиков оставляем как есть, но создадим отдельную колонку-индикатор
games_clean['has_critic_score'] = ~games_clean['critic_score'].isna()
```

```python
# Оценку пользователей оставляем как есть, но создаем индикатор
games_clean['has_user_score'] = ~games_clean['user_score'].isna()
```

```python
# Заполняем пропуски в рейтинге новой категорией
games_clean['rating'] = games_clean['rating'].fillna('Unknown')
```

```python
# Для продаж в Европе и Японии заполняем нулями, так как отсутствие продаж = 0 продаж
games_clean['eu_sales'] = games_clean['eu_sales'].fillna(0)
games_clean['jp_sales'] = games_clean['jp_sales'].fillna(0)
```

```python
# 2. Быстрая нормализация всех текстовых полей
games_clean['genre'] = games_clean['genre'].str.lower().str.strip()
games_clean['platform'] = games_clean['platform'].str.upper().str.strip()
games_clean['rating'] = games_clean['rating'].str.upper().str.strip()
games_clean['name_norm'] = games_clean['name'].str.lower().str.strip()

# 3. Поиск дубликатов (явные и неявные)
print("\nКол-во дубликатов:")

# Явные дубликаты (полные копии)
exact_dups = games_clean.duplicated().sum()
print(f"Явных дубликатов: {exact_dups}")

# Неявные дубликаты (по ключевым полям)
duplicate_mask = games_clean.duplicated(
    subset=['name_norm', 'platform', 'year_of_release'], 
    keep=False
)
print(f"Неявных дубликатов: {duplicate_mask.sum()}")

# 4. Удаление дубликатов 
games_final = games_clean.drop_duplicates(
    subset=['name_norm', 'platform', 'year_of_release']
)

# 5. Итог

print(f"\nРазмер до: {games_clean.shape[0]}")
print(f"Размер после: {games_final.shape[0]}")
deleted = games_clean.shape[0] - games_final.shape[0]
print(f"Удалено записей: {deleted}")
print(f"Процент удалённых данных: {deleted/games_clean.shape[0]*100:.2f}%")
```

## Категоризация данных
    
Проведем категоризацию данных:
- Разделим все игры по оценкам пользователей и выделите такие категории: высокая оценка (от 8 до 10 включительно), средняя оценка (от 3 до 8, не включая правую границу интервала) и низкая оценка (от 0 до 3, не включая правую границу интервала).

```python
# Функция для категоризации пользовательских оценок
def categorize_user_score(score):
    if pd.isna(score):
        return 'Нет оценки'
    elif 8 <= score <= 10:
        return 'Высокая (8-10)'
    elif 3 <= score < 8:
        return 'Средняя (3-8)'
    elif 0 <= score < 3:
        return 'Низкая (0-3)'
    else:
        return 'Некорректная оценка'
```

- Разделим все игры по оценкам критиков и выделите такие категории: высокая оценка (от 80 до 100 включительно), средняя оценка (от 30 до 80, не включая правую границу интервала) и низкая оценка (от 0 до 30, не включая правую границу интервала).

```python
# Функция для категоризации оценок критиков
def categorize_critic_score(score):
    if pd.isna(score):
        return 'Нет оценки'
    elif 80 <= score <= 100:
        return 'Высокая (80-100)'
    elif 30 <= score < 80:
        return 'Средняя (30-80)'
    elif 0 <= score < 30:
        return 'Низкая (0-30)'
    else:
        return 'Некорректная оценка'
```

- После категоризации данных проверим результат: сгруппируем данные по выделенным категориям и посчитаем количество игр в каждой категории.

```python
# Применяем функции для создания новых колонок
df_actual['user_score_category'] = df_actual['user_score'].apply(categorize_user_score)
df_actual['critic_score_category'] = df_actual['critic_score'].apply(categorize_critic_score)
```

```python
# Проверяем результат категоризации
print("\n Категории пользовательских оценок:")
print("-" * 40)
user_categories = df_actual['user_score_category'].value_counts()
for category, count in user_categories.items():
    percentage = (count / len(df_actual)) * 100
    print(f"{category:15} : {count:5d} игр ({percentage:5.1f}%)")

print("\n Категории оценок критиков:")
print("-" * 40)
critic_categories = df_actual['critic_score_category'].value_counts()
for category, count in critic_categories.items():
    percentage = (count / len(df_actual)) * 100
    print(f"{category:15} : {count:5d} игр ({percentage:5.1f}%)")
```
